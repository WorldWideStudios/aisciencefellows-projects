# Real-time chain-of-custody watch
Re-verifies every MMR (append-only), frozen protocols/truths (vs MMR-recorded shas), upstream BioKEA/hardware repo pins, dataset manifests and git state. Breaks are hash-chained events, shown live at `http://<host>:8790/`. Swap `REPO`/`DATASETS` for your own.

```
from baebugs.custody_watch import Watcher
REPO = '.'
DATASETS = '/Volumes/magicBLACKbox/baebugs/daisy/datasets.json'
w = Watcher(REPO, datasets_json=DATASETS, notify=False)
s, changes = w.cycle(remote=True)
print(s['overall'], len(s['checks']), 'checks')
sorted(((c['level'], k, c['reason']) for k, c in s['checks'].items()), key=lambda x: x[0])[:20]
```
