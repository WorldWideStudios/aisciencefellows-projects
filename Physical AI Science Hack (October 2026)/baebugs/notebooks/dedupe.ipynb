# Merkle dedupe with pointer files
One canonical copy per sha256; other copies become `<name>.ptr.json` pointers carrying the hash, canonical path, Merkle index root and inclusion proof. `materialize` restores any pointer (hash-checked). Set `BASE`/`ROOTS` to your data.

```
from baebugs import dedupe as D
BASE = '/Volumes/magicBLACKbox/baebugs'
ROOTS = [BASE + '/datasets']
KEEP = ['datasets/gbif', 'datasets/pseudo_specimens_r3_real']
p = D.plan(ROOTS, BASE, KEEP)
{k: p[k] for k in ('unique','files','pointers','bytes_saved','index_root')}
```

```
# apply to a subset, verify, and restore one
sub = dict(p, items=[i for i in p['items'] if i['path'].startswith('datasets/_rejected_pseudo_v1')][:5])
print(D.apply(sub))
from pathlib import Path
ptrs = sorted(Path(BASE, 'datasets/_rejected_pseudo_v1').rglob('*' + D.SUFFIX))[:5]
[D.verify(x, BASE) for x in ptrs]
```
