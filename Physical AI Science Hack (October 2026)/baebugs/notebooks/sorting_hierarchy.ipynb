# Hierarchical sorting: variable aperture, nested partitions, inserts
One stainless stage steps its opening down through size classes (slot = sorts by thickness, square diagonal-slide = sorts by body width). Each sorting level (size -> morphology -> barcode) is a partition that refines its parent; the tree is a Merkle tree, and leave-one-level-out shows what each level contributes. Swap `objects` for your own specimens.

```
from baebugs.variable_aperture import run, analog_objects, SlotPair
objects = analog_objects()
for mode in ('slot', 'square'):
    r = run(objects, [6, 3, 1.5], pair=SlotPair(mode=mode))
    print(mode, r['status'], {k: sorted({x.split('#')[0] for x in v}) for k, v in r['trays'].items()})
```

```
from baebugs.hierarchy import build, leave_one_level_out
from baebugs.variable_aperture import passes
size = lambda m: next((f'>={w}' for w in (6, 3, 1.5) if not passes(m['dims'], w, 19.05, mode='square')), '<1.5')
shape = lambda m: 'elongate' if sorted(m['dims'])[-1] > 3 * sorted(m['dims'])[1] else 'compact'
levels = [('size', size), ('shape', shape), ('name', lambda m: m['name'])]
tree, novel = build(objects, levels)
print(tree['hash'][:16], len(novel), 'novel')
leave_one_level_out(objects, levels)
```
