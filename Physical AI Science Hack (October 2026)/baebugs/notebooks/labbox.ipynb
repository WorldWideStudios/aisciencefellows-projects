# Lab in a box: BioKEA modules on bench and Opentrons Flex
Modules, printable parts (by STL sha256), arms incl. bring-your-own, upstream repo pins, routes per specimen class, and leave-one-module-out coverage. Add your own arm or part and rebuild.

```
from baebugs.labbox import build, route, lomo, add_arm, BYO_ARM_TEMPLATE
g = build('.')
print(g['root'][:16], len(g['nodes']), 'nodes')
{c: [(s['stage'], s['platform']) for s in route(g, c)['steps']] for c in ('routine', 'vip', 'novel')}
```

```
L = lomo('.')
{m: x for m, x in L['modules'].items() if x['stages_lost']}
```
