# Bench digital twin (VERID-style PyLabRobot tree)
Bench + rail + arm carriage + wood tower (rail-arm slots) + camera station. Every action is checked (reach, swept clearance, fork clearance, order) and hashed; a failed check is a recorded break point. Swap `config/tower.yaml` for measured values.

```
from baebugs import bench_twin as BT, tower_twin as TT
from baebugs.sieve import load_yaml
cfg = load_yaml('tower.yaml')
tw = BT.BenchTwin(cfg=cfg)
print('state_root', tw.state_root())
```

```
plan = TT.plan_tiers(TT.demo_objects()[:60], TT.tower_inventory(), cfg, 2)
acts = TT.actions_for(plan['best'], cfg, TT.initial_state(cfg, TT.tower_inventory()))
res = tw.run_actions(acts)
{k: res[k] for k in res if k in ('status','actions','break_point','mmr_root')}
```

Change `cfg['arm']['reach_mm']` to the measured arm reach and re-run: the break point moves or disappears.
